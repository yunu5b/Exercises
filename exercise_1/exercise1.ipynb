```python
import matplotlib.pyplot as plt
import pandas as pd

drivers = pd.read_csv("../data/Drivers.csv", sep=";")
cosy = drivers["cosy"]
mean = cosy.mean()
std = cosy.std()

print(f"Mean: {mean:.3f}")
print(f"Standard deviation: {std:.3f}")

fig, ax = plt.subplots(figsize=(10, 5))
ax.scatter(range(len(cosy)), cosy, color="tab:blue", alpha=0.75, label="cosy")
ax.axhline(mean, color="tab:red", linestyle="--", linewidth=2,
           label=f"Mean = {mean:.3f}")
ax.axhspan(mean - std, mean + std,
           color="tab:orange", alpha=0.2,
           label=f"Mean ± SD = {mean:.3f} ± {std:.3f}")

ax.set_title("Dot chart of cosy values")
ax.set_xlabel("Observation [-]")
ax.set_ylabel("cosy [min]")
ax.grid(axis="y", alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
```

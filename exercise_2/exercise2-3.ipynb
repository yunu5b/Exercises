```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns  # ty: ignore[unresolved-import]
```

```python
# Daten laden
df = pd.read_csv("../data/Drivers_brand_new.csv")
df.head()

# Fehlende Werte anzeigen
print(df.isna().sum())

# Daten bereinigen, indem Zeilen mit fehlenden Werten in den Spalten "cosy" und "sporty" entfernt werden
df_clean = df.dropna(subset=["cosy", "sporty"])
```

```python
df_clean.describe()
```

```python
# Daten in ein langes Format umwandeln
df_long = df_clean.melt(
    value_vars=["cosy", "sporty"], var_name="Driving style", value_name="Travel time"
)
```

```python
fig, ax = plt.subplots(figsize=(7, 5))

sns.boxplot(
    data=df_long,
    x="Driving style",
    y="Travel time",
    hue="Driving style",
    palette="Set3",
    showfliers=False,
    legend=False,
    ax=ax
)

sns.swarmplot(
    data=df_long,
    x="Driving style",
    y="Travel time",
    hue="Driving style",
    palette="dark:black",
    dodge=False,
    ax=ax
)

ax.set_title("Travel times by driving style")
ax.set_xlabel("Driving style [-]")
ax.set_ylabel("Travel time [min]")

plt.tight_layout()
plt.show()
```

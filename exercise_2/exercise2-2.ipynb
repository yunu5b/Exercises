```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns  # ty: ignore[unresolved-import]

# Daten laden
df = pd.read_csv("../data/Drivers_new.csv")

# Daten vom Wide Format ins Long Format umwandeln
df_long = df.melt(
    value_vars=["cosy", "sporty"], var_name="Driving style", value_name="Travel time"
)

# Gemeinsame Klassenintervalle
bins = 10

# Diagramme erstellen
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Absolute frequency
sns.histplot(
    data=df_long,
    x="Travel time",
    hue="Driving style",
    bins=bins,
    stat="count",
    multiple="layer",
    alpha=0.5,
    edgecolor="black",
    ax=axes[0],
)

axes[0].set_title("Absolute frequency")
axes[0].set_xlabel("Travel time")
axes[0].set_ylabel("Absolute frequency")

# 2. Relative frequency
sns.histplot(
    data=df_long,
    x="Travel time",
    hue="Driving style",
    bins=bins,
    stat="probability",
    multiple="layer",
    alpha=0.5,
    edgecolor="black",
    ax=axes[1],
)

axes[1].set_title("Relative frequency")
axes[1].set_xlabel("Travel time")
axes[1].set_ylabel("Relative frequency")

# 3. Density
sns.histplot(
    data=df_long,
    x="Travel time",
    hue="Driving style",
    bins=bins,
    stat="density",
    common_norm=False,
    multiple="layer",
    alpha=0.5,
    edgecolor="black",
    ax=axes[2],
)

axes[2].set_title("Density")
axes[2].set_xlabel("Travel time")
axes[2].set_ylabel("Density")

fig.suptitle("Travel times: cosy and sporty")
fig.tight_layout()
plt.show()
```
